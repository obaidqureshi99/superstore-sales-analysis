```python
from pathlib import Path

figures_folder = Path("../reports/figures")
figures_folder.mkdir(parents=True, exist_ok=True)

print(figures_folder.resolve())
```

  /////MatplotVisualization

```python
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
plt.figure(figsize=(10, 5))

plt.hist(df_clean["Sales"], bins=90)

plt.xlim(0, 2000)
plt.title("Distribution of Sales (Up to 2000)")
plt.xlabel("Sales Amount")
plt.ylabel("Number of Orders")
plt.savefig(
    figures_folder / "sales_histogram.png",
    dpi=150,
    bbox_inches="tight"
)
plt.show()
```

```python
category_sales = (
    df_clean.groupby("Category")["Sales"]
    .sum()
    .sort_values(ascending=False)
)

plt.figure(figsize=(8, 5))

category_sales.plot(kind="bar")

plt.title("Total Sales by Product Category")
plt.xlabel("Category")
plt.ylabel("Total Sales")
plt.xticks(rotation=0)

plt.tight_layout()
plt.savefig(
    figures_folder / "Categorical_Sales-Barplot.png",
    dpi=150,
    bbox_inches="tight"
)
plt.show()
plt.show()
```

```python
monthly_sales = (
    df_clean.assign(
        Month=df_clean["Order Date"].dt.to_period("M")
    )
    .groupby("Month")["Sales"]
    .sum()
    .sort_index()
)

plt.figure(figsize=(12, 5))

plt.plot(
    monthly_sales.index.astype(str),
    monthly_sales.values,
    marker="o"
)

plt.title("Monthly Sales Trend")
plt.xlabel("Month")
plt.ylabel("Total Sales")
plt.xticks(rotation=45)

plt.tight_layout()
plt.savefig(
    figures_folder / "Monthly_Sale_LineGraph.png",
    dpi=150,
    bbox_inches="tight"
)
plt.show()
```

```python
plt.figure(figsize=(9, 5))

sns.boxplot(
    data=df_clean,
    x="Category",
    y="Profit"
)
plt.title("Profit Distribution by Category")
plt.xlabel("Product Category")
plt.ylabel("Profit per Order")

plt.tight_layout()
plt.savefig(
    figures_folder / "ProfitDistribution_BoxPlot.png",
    dpi=150,
    bbox_inches="tight"
)
plt.show()
plt.show()
```

```python
plt.figure(figsize=(9, 5))

sns.scatterplot(
    data=df_clean,
    x="Discount",
    y="Profit",
    alpha=1
)

plt.title("Relationship Between Discount and Profit")
plt.xlabel("Discount")
plt.ylabel("Profit")

plt.tight_layout()
plt.savefig(
    figures_folder / "ScatterPlot"
    ".png",
    dpi=150,
    bbox_inches="tight"
)
plt.show()
plt.show()
```

```python
numeric_cols = [
    "Sales",
    "Quantity",
    "Discount",
    "Profit",
    "Shipping Days"
]

correlation = df_clean[numeric_cols].corr()

plt.figure(figsize=(8, 6))

sns.heatmap(
    correlation,
    annot=True,
    cmap="coolwarm",
    fmt=".2f",
    vmin=-1,
    vmax=1
)

plt.title("Correlation Between Numerical Variables")

plt.tight_layout()
plt.savefig(
    figures_folder / "HeatMap.png",
    dpi=150,
    bbox_inches="tight"
)
plt.show()
plt.show()
```
